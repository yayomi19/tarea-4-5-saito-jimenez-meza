## **Mercado de autos usados**

Importa el conjunto de datos `neoauto_20240924.csv`
    

```python
import pandas as pd
import numpy as np

base_neoauto = pd.read_csv("neoauto_20240924.csv")
```

1. ¿Cuál es el precio promedio de los autos de cada marca (`item_brand`) publicados en cada subcategoría (`item_category_2`)?

```python
promedio_precios = base_neoauto.groupby(['item_brand', 'item_category_2'])['item_price'].mean().reset_index().round(2)
promedio_precios.rename(columns={'item_price': 'precio_promedio'}, inplace=True)
promedio_precios
```

2. ¿Cuántos autos tienen más de 100,000 kilómetros (`item_km`) pero un precio (`item_price`) menor al promedio general?

```python
precio_promedio = base_neoauto['item_price'].mean()
autos_filtrados = base_neoauto[(base_neoauto['item_km'] > 100000) & (base_neoauto['item_price'] < precio_promedio)]
numero_autos = len(autos_filtrados)
numero_autos
```

3. Encuentra el modelo (`item_name`) más caro y el más barato de cada marca (`item_brand`).

```python
# Agrupar por 'item_brand' y encontrar el índice del auto más caro y más barato
mas_caro = base_neoauto.loc[base_neoauto.groupby('item_brand')['item_price'].idxmax()]
mas_barato = base_neoauto.loc[base_neoauto.groupby('item_brand')['item_price'].idxmin()]

# Imprimir los resultados
for marca in base_neoauto['item_brand'].unique():
    modelo_caro = mas_caro[mas_caro['item_brand'] == marca]
    modelo_barato = mas_barato[mas_barato['item_brand'] == marca]
    
    print(f"Marca: {marca}")
    print(f"  Modelo más caro: {modelo_caro['item_name'].values[0]}, Precio: {modelo_caro['item_price'].values[0]:.2f}")
    print(f"  Modelo más barato: {modelo_barato['item_name'].values[0]}, Precio: {modelo_barato['item_price'].values[0]:.2f}")
    print("-" * 50)
```

4. Encuentra las tres marcas con mayor número de autos financiados por Santander (`item_financed_by`). 

```python
# Filtrar solo los autos financiados por Santander
autos_financiados_santander = base_neoauto[base_neoauto['item_financed_by'] == 'Santander']

# Contar el número de autos por marca
conteo_marcas = autos_financiados_santander['item_brand'].value_counts()

# Seleccionar las tres marcas con más autos financiados
top_3_marcas = conteo_marcas.head(3)

# Imprimir los resultados
print(top_3_marcas)
```

5. Agrupa las publicaciones por tipo de transmisión (`item_transmission`) y calcula la desviación estándar del precio (`item_price`) para cada tipo.  

```python
# Agrupar por tipo de transmisión y calcular la desviación estándar del precio
desviacion_estandar = base_neoauto.groupby('item_transmission')['item_price'].std()

# Imprimir los resultados
print(desviacion_estandar)
```

6. ¿Cuál es el kilometraje promedio de las publicaciones con el tag "Premium" y de las publicaciones con el tag "Como nuevo"?

```python
# Filtrar las publicaciones con el tag "Premium" y "Como nuevo"
premium_autos = base_neoauto[base_neoauto['item_tag'] == 'Premium']
como_nuevo_autos = base_neoauto[base_neoauto['item_tag'] == 'Como nuevo']

# Calcular el kilometraje promedio para cada grupo
km_promedio_premium = premium_autos['item_km'].mean()
km_promedio_como_nuevo = como_nuevo_autos['item_km'].mean()

# Imprimir los resultados
print(f"Kilometraje promedio para autos Premium: {km_promedio_premium:.2f} km")
print(f"Kilometraje promedio para autos Como nuevo: {km_promedio_como_nuevo:.2f} km")
```
