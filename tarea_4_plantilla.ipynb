## **Mercado de autos usados**

Importa el conjunto de datos `neoauto_20240924.csv`
    

```python
import pandas as pd
import numpy as np

base_neoauto = pd.read_csv("neoauto_20240924.csv")
```

1. ¿Cuál es el precio promedio de los autos de cada marca (`item_brand`) publicados en cada subcategoría (`item_category_2`)?

```python
promedio_precios = base_neoauto.groupby(['item_brand', 'item_category_2'])['item_price'].mean().reset_index().round(2)
promedio_precios.rename(columns={'item_price': 'precio_promedio'}, inplace=True)
promedio_precios
```

2. ¿Cuántos autos tienen más de 100,000 kilómetros (`item_km`) pero un precio (`item_price`) menor al promedio general?

```python
precio_promedio = base_neoauto['item_price'].mean()
autos_filtrados = base_neoauto[(base_neoauto['item_km'] > 100000) & (base_neoauto['item_price'] < precio_promedio)]
numero_autos = len(autos_filtrados)
numero_autos
```

3. Encuentra el modelo (`item_name`) más caro y el más barato de cada marca (`item_brand`).

```python
# Agrupar por 'item_brand' y encontrar el índice del auto más caro y más barato
mas_caro = base_neoauto.loc[base_neoauto.groupby('item_brand')['item_price'].idxmax()]
mas_barato = base_neoauto.loc[base_neoauto.groupby('item_brand')['item_price'].idxmin()]

# Imprimir los resultados
for marca in base_neoauto['item_brand'].unique():
    modelo_caro = mas_caro[mas_caro['item_brand'] == marca]
    modelo_barato = mas_barato[mas_barato['item_brand'] == marca]
    
    print(f"Marca: {marca}")
    print(f"  Modelo más caro: {modelo_caro['item_name'].values[0]}, Precio: {modelo_caro['item_price'].values[0]:.2f}")
    print(f"  Modelo más barato: {modelo_barato['item_name'].values[0]}, Precio: {modelo_barato['item_price'].values[0]:.2f}")
    print("-" * 50)
```

## **Evolución económica internacional**

Importa el conjunto de datos de la hoja "Data" del archivo `pwt1001.xlsx`. El diccionario de variables está disponible en la hoja "Legend" del archivo Excel.
    

```python
base_pwt = pd.read_excel('pwt1001.xlsx', sheet_name = "Data")  
base_pwt
```

1. Calcule el PBI per cápita real ajustado por poder de paridad de compra (PPP). Use la variable de PBI que se calcula por gasto y la población. ¿Cuáles son los 10 países top?

```python
# Asegurar que no haya valores nulos en las columnas necesarias
data = base_pwt.dropna(subset=["rgdpe", "pop", "country"]).copy()

# Calcular el PBI per cápita ajustado por PPP
data["pbi_per_capita_ppp"] = (data["rgdpe"] / (data["pop"]))

# Encontrar los 10 países con mayor PBI per cápita
top_10_countries = data.nlargest(10, "pbi_per_capita_ppp")[["country", "pbi_per_capita_ppp"]]

# Mostrar los resultados
top_10_countries.head(10)
```

2. Genera el ránking de los países con el mayor incremento del PBI per cápita real PPP entre los años 1990 y 2019. ¿Dónde se ubica Perú? Para los países que no tengan información para dichos años, puede utilizar datos provenientes de hasta dos años antes o después

```python
# Crear un DataFrame con los años 1990 y 2019 (y tolerar hasta 2 años de diferencia)
data_1990 = data[(data["year"] == 1990)]
data_2019 = data[(data["year"] == 2019)]

# Tomar el valor más cercano a 1990 y 2019 por país
pbi_1990 = data_1990.groupby("country")["pbi_per_capita_ppp"].mean()
pbi_2019 = data_2019.groupby("country")["pbi_per_capita_ppp"].mean()

# Calcular el incremento del PBI per cápita
incremento_pbi = (pbi_2019 - pbi_1990).dropna()

# Generar el ránking
ranking = incremento_pbi.sort_values(ascending=False).reset_index()
ranking.columns = ["country", "incremento_pbi"]

# Encontrar la posición de Perú
posicion_peru = ranking[ranking["country"] == "Peru"].index[0] + 1

# Mostrar los resultados
print("Top 10 países con mayor incremento del PBI per cápita:")
print(ranking.head(10))

print(f"\nLa posición de Perú en el ránking es: {posicion_peru}")
```

3. Para el periodo 2010-2019, calcule el promedio del índice del capital humano. ¿Qué relación existe entre esta variable y el PBI per cápita real PPP calculado previamente? ¿y el número de horas laboradas al año por los trabajadores?

```python
data_2010_2019 = base_pwt[(base_pwt ["year"] >= 2010) & (base_pwt["year"] <= 2019)].dropna(subset = ["hc", "rgdpe", "pop", "avh"])
data_2010_2019["pbi_per_capita_ppp"] = (data_2010_2019["rgdpe"]) / (data_2010_2019["pop"])
prom_hc = data_2010_2019.groupby("country")["hc"].mean()

# Calcular el promedio del PBI per cápita real PPP y horas trabajadas por país
prom_pbi = data_2010_2019.groupby("country")["pbi_per_capita_ppp"].mean()
prom_horas = data_2010_2019.groupby("country")["avh"].mean()

# Unir los datos en un DataFrame
correlacion_data = pd.DataFrame({
    "prom_hc": prom_hc,
    "prom_pbi_per_capita_ppp": prom_pbi,
    "prom_horas": prom_horas
}).dropna()

# Calcular correlaciones
correlacion_pbi = correlacion_data["prom_hc"].corr(correlacion_data["prom_pbi_per_capita_ppp"])
correlacion_horas = correlacion_data["prom_hc"].corr(correlacion_data["prom_horas"])

# Resultados
print(f"Correlación entre el índice de capital humano y el PBI per cápita real PPP: {correlacion_pbi:.2f}")
print(f"Correlación entre el índice de capital humano y las horas trabajadas: {correlacion_horas:.2f}")
```

7.  Genere una variable que ordene a los países según el porcetaje de formación bruta de capital del 2019. Luego, establezca diez grupos a manera de deciles. ¿En qué decil se ubica Péru? Vuelva a calcular usando la información de 1990. ¿El Perú se mantiene en el mismo décil o cambia?

```python
# Filtrar los datos para 2019 y 1990
base_2019 = base_pwt[base_pwt['year'] == 2019]
base_1990 = base_pwt[base_pwt['year'] == 1990]

# Ordenar por el porcentaje de formación bruta de capital 
base_2019_sorted = base_2019.sort_values(by='csh_i', ascending=False)

# Crear deciles para 2019
base_2019_sorted['decile_2019'] = pd.qcut(base_2019_sorted['csh_i'], 10, labels=False) + 1

# Encontrar el decil de Perú en 2019
peru_decil_2019 = base_2019_sorted[base_2019_sorted['country'] == 'Peru']['decile_2019'].values[0]

# Repetir el cálculo para 1990
base_1990_sorted = base_1990.sort_values(by='csh_i', ascending=False)
base_1990_sorted['decile_1990'] = pd.qcut(base_1990_sorted['csh_i'], 10, labels=False) + 1

# Encontrar el decil de Perú en 1990
peru_decil_1990 = base_1990_sorted[base_1990_sorted['country'] == 'Peru']['decile_1990'].values[0]

# Mostrar los resultados
print(f"El decil de Perú en 2019 es: {peru_decil_2019}")
print(f"El decil de Perú en 1990 es: {peru_decil_1990}")
```

8. Filtra todas las transacciones de tipo "Debit" en las que el monto sea mayor a 200, y calcula el monto total de esas transacciones. Además, determina cuántos clientes únicos realizaron esas transacciones.

```python
data_debit_mayor_200 = base_finance[(base_finance["Transaction_Type"] == "Debit") & (base_finance["Amount"] > 200)]
monto_total = data_debit_mayor_200["Amount"].sum()
clientes_unicos = data_debit_mayor_200["Customer_ID"].nunique()

print(f"Monto total de las transacciones 'Debit' mayores a 200: {monto_total}")
print(f"Cantidad de clientes únicos que realizaron esas transacciones: {clientes_unicos}")
```

9. Identifica las transacciones con valores nulos en la columna (`Amount`) . Rellena esos valores con la mediana de los montos por tipo de transacción (`Transaction_Type`) y luego calcula el monto total de transacciones por región (`region`) .

```python
# 1. Identificar las transacciones con valores nulos en la columna 'Amount'
transacciones_nulas = base_finance[base_finance['Amount'].isna()]

# 2. Rellenar los valores nulos con la mediana de 'Amount' por tipo de transacción ('Transaction_Type')
base_finance['Amount'] = base_finance.groupby('Transaction_Type')['Amount'].transform(lambda x: x.fillna(x.median()))

# 3. Calcular el monto total de las transacciones por región
monto_total_por_region = base_finance.groupby('Region')['Amount'].sum()

# Mostrar el monto total de transacciones por región
print("\nMonto total de transacciones por región:")
print(monto_total_por_region)
```

10. Busca transacciones duplicadas considerando las columnas (`Customer_ID`) y (`Date`). Elimina los duplicados manteniendo solo la primera ocurrencia. Luego, agrupa las transacciones restantes por (`Customer_ID`) y calcula el monto promedio y la cantidad de transacciones por cliente.

```python
# 1. Buscar transacciones duplicadas considerando las columnas 'Customer_ID' y 'Date'
transacciones_sin_duplicados = base_finance.drop_duplicates(subset=['Customer_ID', 'Date'], keep='first')

# 2. Agrupar las transacciones restantes por 'Customer_ID'
# Calcular el monto promedio y la cantidad de transacciones por cliente
resumen_cliente = transacciones_sin_duplicados.groupby('Customer_ID').agg(
    monto_promedio=('Amount', 'mean'),
    cantidad_transacciones=('Transaction_ID', 'count')
)

# Mostrar el resultado
print(resumen_cliente)
```
